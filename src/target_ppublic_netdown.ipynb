```python
import pandas as pd
```

```python
df_ppublic = pd.read_csv('../dat/tmp/tsa_ppublic.csv', delimiter='\t').set_index('tsa')
```

```python
df_ppublic.head()
```

```python
df_targets = pd.read_csv('../dat/targets.csv').set_index('tsa')
```

```python
vcut_sum_target = df_targets['vcut'].sum()
```

```python
df_targets = df_targets.join(df_ppublic)
```

```python
df_targets.head()
```

```python
for c in ['vcut', 'abrn']:
    df_targets[c] = (df_targets[c] * df_targets['ppublic']).astype(int)
```

```python
df_targets = df_targets.drop('ppublic', 1)
```

```python
df_targets.head()
```

```python
vcutp1_sum_target =  54815521 
vcutp1_sum = df_targets.reset_index().set_index(['tsa', 'period'])['vcut'].loc[:, 1].sum() * 0.1
```

```python
vcutp1_sum_target, vcutp1_sum
```

```python
df_targets['vcut'] = (df_targets['vcut'] * vcut_sum_target / df_targets['vcut'].sum()).astype(int)
```
